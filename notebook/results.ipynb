# Analysis of groundtruth


```python
folder_gt = r"C:\Users\tizia\Documents\Curie\Python for Morgane\Groundtruth"
name_file_gt = r"TT_divisiontimes.xlsx"
name_sheet_list = ["MB07082025", "MB03102025"]
file_path = os.path.join(folder_gt, name_file_gt)

sheet_config = {
    "MB03102025": {
        "regex": r"\(([^)]+)\)",
        "mapping": {"HSC_2000":"HSC", "MPP_epp": "MPP", "LMPP_epp":"LMPP", "HPC_3000":"HPC"}, 
        "order": ["HSC", "MPP", "LMPP", "HPC"],
        "palette": {"HSC":  "#F4A896",  
                "MPP":  "#E8604C", 
                "LMPP": "#C0392B",  
                "HPC":  "#7B3F00"}},
       
    "MB07082025": {
        "regex": r"(TR\d+)",
        "mapping": {"TR1":"BMS-265246","TR3":"Venetoclax","TR5":"JQ-1","TR7":"Palbociclib","TR9":"Control"},
        "order": ["BMS-265246","Venetoclax","JQ-1","Palbociclib","Control"],
        "palette": {
        "BMS-265246": "#4C9BE8",  
        "Venetoclax": "#5B8FD4",  
        "JQ-1": "#3A6FAF",  
        "Palbociclib": "#5BBCD6",  
        "Control": "#6B8FA8"
    }
    },
}

var_boxplot_list = ["T1", "T2.1", "T2.2", "T3.1", "T3.2", "T3.3", "T3.4"]
scatter_pairs = [("T1","T2.1"),("T2.1","T2.2"),("T3.1","T3.2"),("T3.2","T3.3"),("T3.3","T3.4")]
division_groups = [["T2.1","T2.2"], ["T3.1","T3.2","T3.3","T3.4"]]

seed = 42
rng = np.random.default_rng(seed)

for name_sheet in name_sheet_list:
    data = pd.read_excel(file_path, sheet_name=name_sheet)
    
    config = sheet_config[name_sheet]
    data["Condition"] = data["Name"].str.extract(config["regex"], expand=False)
    if config["mapping"]:
        data["Condition"] = data["Condition"].replace(config["mapping"])
    
    ordered_conditions = config["order"] or sorted(data["Condition"].dropna().unique())
    palette = config["palette"] or None
    # statistical tests
    for var_boxplot in var_boxplot_list:
        print(data.groupby("Condition")[var_boxplot].describe())
        analyze_gt(data, name_sheet, ordered_conditions, var_boxplot, palette)

    # randomize
    data_random = data.copy()
    for group in division_groups:
        data_random = randomize_rowwise(data_random, [group], seed=seed)

    # scatter + spearman
    for x, y in scatter_pairs:
        plot_scatter(data_random, x, y, ordered_conditions)
        compute_spearman(data_random, x, y, ordered_conditions)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_histograms_indices(global_flat_df, path_save):
    scale = 4
    plt.rcParams.update({
            'font.family': 'sans-serif',
            'font.size': 6 * scale,
            'axes.labelsize': 7 * scale,
            'axes.titlesize': 7 * scale, # for axes
            'figure.titlesize': 7 * scale, # whole figure
            'xtick.labelsize': 4 * scale,
            'ytick.labelsize': 4 * scale,
            'legend.fontsize': 4 * scale,
        })


    # only OCI
    #df_oci = global_flat_df[global_flat_df["CellType"] == "oci"].copy()
    # Everything
    df_oci = global_flat_df.copy()
    df_oci = df_oci[df_oci["T1"].notna()]

    fig, axes = plt.subplots(1, 3, figsize=(4.5*scale, 1.5*scale))

    bin_width = 0.1
    print(len(df_oci))

    # Panel 1: OscIdx (full experiment)
    osc_full_data = df_oci["Osc_index"].dropna()
    print(len(osc_full_data))
    bins_osc_full = np.arange(0, osc_full_data.max() + bin_width, bin_width)

    axes[0].hist(osc_full_data, bins=bins_osc_full, color="#5C2685")
    #axes[0].axvline(0.2, color="red", linestyle="--", label="Threshold = 0.2")
    axes[0].set_xlabel("OscIdx")
    axes[0].set_ylabel("Number of microwells")
    axes[0].legend()

    # Panel 2: OscIdx8
    osc_data = df_oci["Osc_index-9"].dropna()
    print(len(osc_data))
    bins_osc = np.arange(0, osc_data.max() + bin_width, bin_width)

    axes[1].hist(osc_data, bins=bins_osc, color="#7A34AC")
    axes[1].axvline(0.2, color="red", linestyle="--", label="Threshold = 0.2")
    axes[1].set_xlabel("OscIdx8")
    axes[1].legend()

    # Panel 3: TrustIdx (no negative values)
    #trust_data = df_oci[df_oci["Trust_index"] >= 0]["Trust_index"].dropna()
    #bins_trust = np.arange(0, trust_data.max() + bin_width, bin_width)

    trust_data = df_oci["Trust_index"].dropna()
    bin_min = np.floor(trust_data.min() / bin_width) * bin_width  
    bins_trust = np.arange(bin_min, trust_data.max() + bin_width, bin_width)

    axes[2].hist(trust_data, bins=bins_trust, color="#B19CD9")
    axes[2].axvline(0.8, color="red", linestyle="--", label="Threshold = 0.8")
    axes[2].set_xlabel("TrustIdx")
    axes[2].legend()

    plt.tight_layout()
    plt.savefig(f"{path_save}/fig_indexes_quality_distribution.svg", bbox_inches="tight")
    plt.savefig(f"{path_save}/fig_indexes_quality_distribution.png", dpi=300, bbox_inches="tight")
    plt.show()

plot_histograms_indices(global_flat_df, path_save = "Figures")
```
